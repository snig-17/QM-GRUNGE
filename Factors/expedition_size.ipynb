```python
import pandas as pd

from google.colab import files
uploaded = files.upload()


filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

df_filtered = df[df['Year'] >= 1990]

plt.figure(figsize=(10, 6))
sns.scatterplot(x='Year', y='group_size', data=df_filtered)
plt.title('Expedition Size Over the Years')
plt.xlabel('Year')
plt.ylabel('Expedition Size')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

ycol = "group_size"

# Drop rows where 'expedition_size' is NaN, and filter for years after 1990
df_cleaned = df.dropna(subset=[ycol, 'Year'])
df_cleaned = df_cleaned[df_cleaned['Year'] > 1990]

plt.figure(figsize=(14,6))
plt.hist2d(df_cleaned["Year"], df_cleaned[ycol], bins=[36, 50], cmap='BuPu')  # Changed colormap to 'viridis'
plt.title("Expedition Size vs Year (Density) (After 1990)")
plt.xlabel("Year")
plt.ylabel("Expedition Size")
plt.colorbar(label="Count")
plt.grid(False)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

ycol = "group_size"

# Filter data for years after 1990
df_filtered_post_1990 = df[df['Year'] > 1990]

x = df_filtered_post_1990["Year"].to_numpy()
y = df_filtered_post_1990[ycol].to_numpy()

# jitter the x slightly so points don’t stack exactly
x_jitter = x + np.random.uniform(-0.15, 0.15, size=len(x))

plt.figure(figsize=(14,6))
plt.scatter(x_jitter, y, alpha=0.25, s=15)
plt.title("Expedition Size Over the Years (Jittered Scatter, After 1990)")
plt.xlabel("Year")
plt.ylabel("Expedition Size")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import MultipleLocator

# Filter data for years after 1990
df_filtered_post_1990 = df[df['Year'] > 1990]

plt.figure(figsize=(10, 6))
sns.scatterplot(x='group_size', y='success_rate', data=df_filtered_post_1990, alpha=0.6)
plt.title('Expedition Group Size vs. Success Rate (After 1990)')
plt.xlabel('Group Size')
plt.ylabel('Summit Success Rate')



plt.grid(True, alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter data for years after 1990
df_filtered_post_1990 = df[df['Year'] > 1990]

plt.figure(figsize=(10, 6))
plt.hexbin(x=df_filtered_post_1990['group_size'], y=df_filtered_post_1990['success_rate'], gridsize=30, cmap='Blues')
plt.colorbar(label='Count')
plt.title('Expedition Group Size vs. Success Rate (Hexbin Plot, After 1990)')
plt.xlabel('Group Size')
plt.ylabel('Summit Success Rate')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Create a copy to avoid modifying the original filtered DataFrame directly
df_plot = df[df['Year'] > 1990].copy()

df_plot['success_rate'] = pd.to_numeric(df_plot['success_rate'], errors='coerce')
df_plot['success_rate'] = df_plot['success_rate'].clip(upper=1)

# Display the head of the new DataFrame to confirm changes
display(df_plot.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter data for years after 1990 and use the cleaned df_plot
# df_filtered_post_1990 = df[df['Year'] > 1990] # This line is no longer needed directly

plt.figure(figsize=(10, 6))
sns.scatterplot(x='success_rate', y='group_size', data=df_plot, alpha=0.6)
plt.title('Expedition Summit Success Rate vs. Group Size (After 1990)')
plt.xlabel('Summit Success Rate')
plt.ylabel('Group Size')
plt.grid(True, alpha=0.3)
plt.show()
```
